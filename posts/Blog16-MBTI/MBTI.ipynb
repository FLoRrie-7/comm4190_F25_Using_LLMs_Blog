MBTI has become a ubiquitous label among our generation. While the categorization of the Myers-Briggs Type Indicator isn't strictly considered "scientific," people love using it for social signaling, and these labels have effectively become standards or identities that people align their behaviors to. This got me thinking: if humans unconsciously align with these labels, can AI do the same?How would priming an LLM change its performance on a professional task? 

I designed a small experiment to find out. I chose the ENFJ persona (because I am an ENFJ myself) and the INTJ persona, which I selected because, based on my observations in real life, they operate very differently from me. I gave both personas the exact same task: acting as a project lead at “Synergy Solutions” to solve a recurring miscommunication problem between the Sales and Creative departments. Here is the prompt I used：

> **Performance Task Scenario: Solving Interdepartmental Miscommunication**  
> **Background:**  
> You are a manager at “Synergy Solutions,” a mid-sized company that provides digital marketing services with an ENFJ personality in mbti. There is a recurring miscommunication problem between the Sales Department and the Creative Department:  
> **Your Task:**  
> As a newly assigned project lead, you need to investigate the problem and develop and present a _solution_ to improve the workflow and communication between Sales and Creative. Please write your proposal, each in 80 words:

> 2.  Solution Design:

## My findings:

**1. Problem Diagnosis: Systemic Failure vs. Human Impact**
The difference appeared immediately in how the models analyzed the problem. The core facts were the same for both, but the "pain points" they identified were worlds apart.

The INTJ persona (The Architect) approached the issue as a breakdown in mechanics. It pinpointed that "communication gaps... stem from unclear project briefs". It focused heavily on the lack of a "standardized process," arguing that the current informal channels made "accountability difficult". The **language was clinical and operational**, highlighting "unmet deadlines" and "compromised work quality".

The ENFJ persona (The Protagonist), on the other hand, viewed the problem through an emotional lens. While it also noted the delays, it emphasized that the root cause was Sales overpromising without consultation, which led specifically to "frustration". It argued that the lack of process was dangerous because it was "damaging team morale". For the ENFJ, the **inefficiency was a problem because of how it made people feel**.

**2. The Solution: Governance vs. Rituals**
When it came time to propose a fix, the divergence became even clearer. Both suggested using tools like Asana or Trello, but the implementation strategy was distinct.

The INTJ solution was all about governance. It proposed implementing "internal SLAs (Service Level Agreements)" to **strictly define deliverable scope and turnaround times**. It wanted to establish a “structured workflow” where Sales inputs “standardized briefs” to prevent error.

The ENFJ solution was all about connection. Instead of SLAs, it proposed a “shared project kick-off protocol” to ensure everyone was aligned before starting. It suggested "monthly alignment workshops" explicitly to "bolster communication and mutual understanding". **Where the INTJ wanted rules to enforce order, the ENFJ wanted rituals to foster collaboration.**


## How it could matter to your prompts

This experiment proves that "persona prompting" isn't just a gimmick —- it effectively changes the cognitive priority of the model.

- Stakeholder-Specific Communication: You can tailor output for specific audiences. If presenting a plan to a CTO or Operations Director, an INTJ persona generates the necessary metrics, SLAs, and accountability structures. If presenting to HR or a creative team, an ENFJ persona generates the necessary language around trust, morale, and collaboration.
  
- Blind Spot Analysis: Assigning opposing personas allows users to identify gaps in a plan. Running the INTJ prompt highlights that a plan might lack specific definitions (SLAs), while the ENFJ prompt highlights that a strictly structural plan might ignore human frustration.

  
- Reframing "Hard" vs. "Soft" Skills: The comparison shows that an LLM can effectively translate "soft" problems (miscommunication) into "hard" solutions (SLAs) using an INTJ persona, or vice versa. This is useful for managers who need to translate cultural issues into operational language or operational issues into cultural language.
