After reading Study 1, my first instinct was: Maybe we can just prompt it better. Give GPT-4 explicit instructions: “Use active learning. Ask questions. Encourage collaboration. Use Socratic principles to guide the students.”

Study 2 in the same paper is essentially a large-scale test of that instinct. And the results are slightly unsettling. The authors ask a second question: If we provide more educationally specific inputs, can we improve GPT-4’s pedagogical alignment?

Their resutls showed that: **You can make the model more controllable and predictable—but alignment quality may still resist change, especially for complex pedagogies.**


## Study 2: “Teaching design” at scale + a metric called PAD

In Study 2, the researchers use the OpenAI API to generate thousands of course designs in two stages:

**Stage 1 (baseline):**  course attributes like topic, class size, engagement methods.

**Stage 2 (enriched):** add a “skills list” explicitly tied to learning theories (e.g., memory vs analysis vs collaboration).

They quantify alignment using Pedagogical Alignment Distance (PAD), ranging from -1 (misaligned) to +1 (aligned).

**The surprising part: “more guidance” didn’t fix overall alignment**

Even with enriched, theory-linked inputs, the paper reports that although Behaviorism improved in Stage 2 (ΔPAD = +0.11), Active learning and Social learning remained negative, with reported values like PAD = -0.17 (Active) and PAD = -0.29 (Social)

So the model became **more responsive to the instruction variable, but not necessarily more capable of executing the deeper pedagogy**. The authors also show that the added “skills” variable becomes the most influential predictor.



## What this means for product builders: controllability ≠ pedagogy

This is where the paper shifts from “research result” to “product warning.”
If you build an education app and your solution to pedagogical quality is “just prompt it with better instructions,” you may get more consistent outputs, easier-to-debug behavior, cleaner A/B results, but still fail at: social learning facilitation, inquiry-driven scaffolding, knowing when not to give the answer. In other words, the paper suggests that prompting can increase predictability without increasing pedagogical intelligence.

This paper reframes “alignment in education” as something more concrete than values or safety: it’s alignment to teaching theories—and that alignment appears to have stubborn defaults.

***Fine-tuning for pedagogical intent as a path forward?***

The paper suggests that fine-tuning is the *“change the defaults”* route —but it depends on high-quality teaching dialogue data.

Two related directions in the literature are worth watching:

- *Pedagogical alignment via preference learning (LHP)*, which argues fine-tuning can push models away from “give the answer” behavior.
  
- *Fine-grained pedagogical intent annotation*, which shows that more detailed intent labels can improve tutoring response quality.

## Final Thoughts

Education is broad. Over-specializing a model is risky because teachers constantly shift across domains, goals, and student needs. A “perfect Socratic tutor” that becomes brittle or forgets general skills is not a win.
