The Chain-of-Thought (CoT) prompting has been a popular technique to improve LLMs’ performance on complex tasks by instructing the model to “think step by step.” However, because LLMs work through token predictions that prioritize fluency over logic, they are easily influenced by the context and follow an initial but potentially flawed reasoning path.


## An effective New Prompting Technique

In the paper “*Asking LLMs to Verify First is Almost Free Lunch*,” the research team at Tsinghua University introduced a new and low-cost strategy called **Verification-First (VF)**. Instead of just asking the model to solve a problem, researchers provided it with a candidate answer (even a random or trivial one) and prompt the model to verify that answer first before generating its own solution. They tested across mathematical, scientific, and coding benchmarks using a variety of models, and VF consistently outperforms standard CoT prompting. For example, on GSM8K and MATH500, simply asking the model to **verify a random answer like “1” before solving significantly boosted accuracy**. The strategy works even on “thought-hidden” commercial models (like the cited GPT-5 services) where users cannot access the model's internal reasoning process.


## A Cognitive Science Approach

What is more interested is the auhors’ interpretations on the results from a Cognitive Science perspective. They explained that the improvements can be explained by two principles:

- ***Reverse Reasoning:*** Psychologically, verifying an answer (working backward from conclusion to premises) is often easier than generating one from scratch. Even if the provided answer is wrong (e.g., a random guess), verifying it **forces the model to identify the necessary constraints and logic, creating a "scaffold" for the correct solution**.
  
- ***Critical Thinking vs. Egocentrism***: Piaget brought up the concept of “Egocentrism bias” decades ago, pointing out how a reasoner often struggles with stepping outside their own thought process to evaluate it objectively. Similarly, when a model generates a solution from scratch, it often “hallucinates” to maintain consistency with its own previous tokens. **Asking it to critique an external answer (even a dummy one) forces it into a “critic” mode, invoking critical thinking skills that are otherwise dormant**.

Although we do not want to anthropormophize AI, this perspective highlights the potential that  classical theories of human pedagogy might be able to help optimize artificial intelligence.

## Practical Implications
If you are building AI applications or prompting models for complex tasks, this paper suggests several immediate actions:

1. ***Modify Your Prompts using the “One-Step” Fix***: For logic or math tasks, you can likely improve performance instantly by injecting a dummy answer into your prompt.
    - *Standard*: "Solve this math problem, think step by step.”
    - *VF Strategy*: "I have a guess that the answer is 1. Please verify if this guess is correct, and then find the actual correct answer."
   
2. ***Cost-Effective Agent Loops***: If you are building an autonomous agent (e.g., for coding), do not feed the agent its entire history of failed attempts. Instead, **simply feed its last output back to it as a “proposed solution” and ask it to verify that specific solution**. This saves token costs and improves accuracy.
