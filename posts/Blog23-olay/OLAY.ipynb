In the book Unmasking AI, Dr. Joy Buolamwini shares how, as her work in exposing algorithmic bias became more socially recognized, companies started reaching out. One of the most interesting collaborations was with OLAY, which approached her for a partnership involving an algorithm audit on their AI identification and recommendation system.

It is good to see big companies, even as part of their branding strategies, get involved in such campaigns. It effectively raises social awareness, making the public realize that algorithmic bias isn't an abstract concept—it’s in their everyday skincare routine. We need more research work to go public, and we certainly need more companies to move beyond lip service and show concrete action.


## The Problem: Bias is More Than Skin Deep

The collaboration, aptly named "Decode the Bias in Beauty," was necessary because Buolamwini’s previous research demonstrated that many AI systems struggle with accuracy when encountering darker skin tones, often due to dataset deficiencies. The target of the audit was OLAY's Skin Advisor tool, which uses AI to analyze a user’s facial image and recommend personalized products.

The question wasn't if the system was biased, but how and where.

The audit, conducted by Buolamwini's consultancy ORCAA, confirmed the critique: the AI model produced significantly lower confidence scores for darker skin tones compared to lighter ones. In the language of AI, this means the model was less certain and, therefore, less reliable when advising consumers of color. This finding perfectly validates the core thesis of algorithmic justice: a system can perform "well enough" overall while failing miserably at the intersection of race and technology.


## The Insight: From "Good Enough" to Accountable

The true insight here is not just that bias was found—that was predictable. The breakthrough lies in the company's response. OLAY didn't just quietly fix the problem; they publicly released the audit results and committed to fixing the underlying data disparity.

This led to Project **D.A.T.A. (Driving Accurate Representation in Testing and Analysis)**, a commitment to dramatically increase the diversity of their training datasets. This moves the discussion from ethical theory to data accountability. For a brand rooted in mass-market consumer confidence, admitting that their digital cornerstone was unreliable for a significant segment of their customers is a powerful step. It forces them to look beyond the superficial "overall accuracy" metric and confront the ethical implications of their "ground truth."


## The Call to Action: Making Change Structural

While the public nature of the partnership is great for branding, the real impact lies in whether this moment catalyzes structural change across the industry. The OLAY partnership offers a hopeful vision: a world where corporate self-interest in good branding aligns with the ethical imperative of algorithmic justice. It is a necessary first step towards ensuring that the tools we build—whether they recommend skincare or determine job eligibility—do not leave anyone behind.
