In the book Unmasking AI, the author connects a core concept in legal theory—intersectionality—to a deep flaw in machine learning. The book draws on the work of legal scholar Kimberlé Crenshaw to explain **“multi-axis discrimination”** and shows how AI risks repeating the exact same mistakes.

## What is Multi-Axis Discrimination? 

Coined by Crenshaw, intersectionality describes how individuals can face multiple, overlapping forms of discrimination, like racism and sexism. These factors don't exist in isolation; they can combine to create unique forms of disadvantage.
When systems only look for “racism” or “sexism” (a single axis), they often completely miss the compound discrimination faced by those at the intersection, such as women of color.
The book provides a stark hiring example: A company “proves” it isn’t racist by pointing to the Black men it hired, and “proves” it isn’t sexist by pointing to the white women it hired. By looking at race and gender separately, the system meets the law. But it completely overlooks the Black women, Asian women, or other women of color who are systematically excluded. **If they don’t get hired, they would have no legal basis to rederess**. 


## The AI Connection: The Tyranny of the Single Metric

This is where the author connects this legal blind spot to machine learning. AI models are often judged by a single, “overall" performance metric, like “95% accuracy” in a predictive task. 
But this single number is just like the company’s flawed hiring defense. It hides who the model is failing.
Let's apply this to a facial recognition system. An “overall accuracy” of 95% might break down like this:
- Performance on white men: 99.8%
- Performance on white women: 97%
- Performance on Black men: 85%
- Performance on Black women: 70%
  
The “overall” score looks great only because it's **weighted by high performance on the dominant groups**. It masks the system's dangerous failures at the intersections. The system isn’t “95% accurate”; it's a system that works for some and fails, often harmfully, for others.

## The Impact on Educational Equity

This reminded me of a machine learning project in education that I did. The “single metric” problem has dangerous and practical implications for education equity, where AI is increasingly used to make high-stakes decisions.

Schools already lean heavily on single metrics: GPA, standardized test scores, “on track” flags in student dashboards. As AI tools enter classrooms—adaptive learning platforms, early-warning systems, automated essay scoring, “at-risk” prediction models—they often replicate that same logic in code. A model might boast “92% accuracy in predicting who will pass the course,” but that number can obscure which students it systematically gets wrong.

Imagine a district using an AI system to identify students who need extra support. The model performs well “on average,” but breaks down like this:
- High accuracy for white, middle-class students
- Lower accuracy for multilingual learners
- Even worse accuracy for, say, recently immigrated girls of color in under-resourced schools
  
On paper, the system looks like a success. In practice, it might **consistently under-flag some groups** (denying them needed support) or **over-flag them** (subjecting them to extra surveillance, remediation, or stigma). Just as in the hiring example, the district can point to data showing “we support girls” and “we support students of color” in aggregate—while missing those at the intersection who are misclassified, misunderstood, or left behind.


## The Path Forward

For educators and designers, the lesson is clear: it’s not enough to ask, “Does this AI work for students?” We have to ask, “For which students, under what conditions, and at what intersections of race, gender, language, disability, and class?” That means disaggregating performance metrics, involving students and families from marginalized groups in the design process, and treating “overall success” as a starting point, not a finish line.

If we don’t, AI in education risks becoming just another polished dashboard that tells a comforting story about improvement—while quietly reproducing the same old patterns of inequity, now with the authority of an algorithm behind them.
