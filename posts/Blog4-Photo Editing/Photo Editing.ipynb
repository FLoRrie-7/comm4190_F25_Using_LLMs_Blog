## Let the GPT provide aesthetic suggestions
Besides schoolwork, I have been working on editing my birthday portraits diligently, and suddenly became curious about how an LLM could help. Photo editing requires emotional perception and a keen aesthetic sense. Would a connection-based model like GPT be capable of such a task?

To test its ability, I selected three photos, each with a slightly different color tone. I wrote the following prompt:


>*These are some portraits I took during my birthday. You are an experienced photographer who has strong expertise and a good sense of aesthetic judgment. Please give me detailed suggestions on how to edit or do color adjustment on these raw photos to enhance them and make them more appealing and artistic. Please also explain the rationales behind your suggestions.*

## A number-extensive response
GPT 5’s thinking mode produced a response that seems to be really comprehensive. It startw with a global look to unify the set of the three photos, and identify the goal should be “dreamy, romantic ‘buttery yellow’ palettee.” Based on this goal, he gave out the exact statistics on the references:


```
**White balance**
*Outdoor shots*: push a touch warmer and magenta to counter green foliage and blue skylight on skin. Temp +300 to +600K, Tint +5 to +12.
*Indoor window shot*: neutral-warm but not orange. Temp –100 to +150K from camera, Tint +3 to +7.

**Tone Curve**
Subtle S-curve*:Lift shadows slightly (point curve floor to ~4–6).Pull mid-highlight just a hair up. Optionally lift extreme highlights minimally to keep porcelain statue detail.

**Presence**
*Texture*: –5 to –15 on skin (via mask), +5 to +10 on hair/eyes/dress stitching.
*Clarity*: 0 to +8 overall; avoid high clarity on skin.
*Dehaze*: –2 to +3 only; too much looks crunchy.
```

After this unified overview, it then provided specific "recipes" for each photo, with **extensive numbers and metrices** just like above. Finally, it listed export suggestions to preserve the final look, which I thought was a nice, thoughtful touch. To conclude, it invited more personalized adjustments, likely a learned behavior from its training process to encourage further interaction.

##The final product
By this point, however, I was already overwhelmed by the sheer volume of numbers. In a natural interaction with a human photographer, they would never recite such specific statistics right away. They would **show you the effects visually**, because human senses don't directly process numbers in such an aesthetic and perceptual task. So, I asked it to apply these settings directly to the photos.

## A different approach than humans
The process of editing a photo is often a **trial-and-error** process for human. Even an experienced editor will experiment with different settings, constantly evaluating and comparing the results, and then continuously adjusting based on their experience and intuition. The exact numbers given by the LLM can be a helpful initial starting point, but **lacking the ability for holistic, emotional judgment**, it can only provide solutions based on its training data and wait for a human's instructions to make iterative adjustments.

## Final thoughts
This also makes me wonder—since all its solutions are derived from associative data, can it only produce **"average"** artwork? Can it create **novel and bold aesthetics through genuine experimentation**? And how might different prompting strategies affect that outcome?
