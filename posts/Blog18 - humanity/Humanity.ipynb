The author of Unmasking AI, Dr. Joy Buolamwini, is far more than just a computer scientist. While she possesses a deep technical background, her work is profoundly shaped by the reflections of her lived experience. It was a specific failure of technology—an AI tool that could not detect her face until she put on a white mask—that forced her to reflect on the implications of her own dark skin in a digital world. This moment motivated her to investigate deeper into the algorithmic biases, such as race and gender, that exist within the machine learning world.

These experiences and humanities-driven perspectives gave her a clear aim: to utilize her technical skills for real-world impact. She has since presented her work to tech giants like IBM, pushed for policy changes in academia and government, delivered a TED Talk, and even wrote poems to convey her findings in an artistic, compelling way. She demonstrates that technical skill, when guided by deep human conviction, becomes a tool for justice.


## An Innovative Education/Internship Program to Cultivate Technicians with Purposes

Interestingly, a recent news story resonates with this idea. The data analytics company Palantir recently launched a "Meritocracy Fellowship," offering high-school graduates with exceptional coding talent the chance to skip college and work at the company. The program is incredibly competitive, with an admission rate of just 4.4%. However, follow-up reports revealed a surprising twist: these interns spent their first month not writing code, but studying Plato’s Republic, the leadership styles of Lincoln and Churchill, and the foundations of Western civilization.

This curriculum exists because Palantir works primarily with high-stakes clients like the US Department of Defense and various intelligence agencies. Their rationale is that engineers cannot simply understand how to write code; they must also understand why they are writing it and who they are serving. They are effectively criticizing an undergraduate education system that often produces technically proficient but "aimless" engineers who lack a grasp of the values they are vouching for.

## The "Why" is the New Syntax

This convergence of stories—one from an algorithmic justice activist and one from a defense contractor—signals a shift in what defines a "skilled" engineer. In an era where AI tools like GitHub Copilot and ChatGPT can increasingly handle the "how" of coding (syntax, boilerplate, and optimization), the human value-add is shifting entirely to the "why." The ability to write an efficient algorithm is becoming a commodity; the ability to discern whether that algorithm should be written, or to predict its downstream societal impacts, is becoming a luxury skill. We are moving toward a future where the most valuable engineers will not be the ones who code the fastest, but the ones who can anchor their code in a robust moral and historical framework.
