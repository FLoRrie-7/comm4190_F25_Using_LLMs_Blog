My interview with Gemini last time was fun, feeling much like a childhood pretend play. The language it adopted as a high schooler felt authentic, and its subsequent role as a teacher demonstrated a truly "pedagogical" voice and mannerisms. This made me wonder: Beyond merely affecting tone, does assigning a specific persona to an LLM truly enhance its task performance? And if so, in what ways?

Conducting a thorough experiment myself would be challenging, so I turned to the resesarch papers for a comprehensive and validated answer. Here’s what I found:

## Assigning a persona to an LLM can have a multifaceted impact

**On Reasoning Ability:**

- **Benefits:** A 2023 study found that role-play prompting can act as an implicit, as well as a better "Chain-of-Thought" (CoT) trigger than CoT instruction alone. It **guides the model to break down complex problems into a series of intermediate reasoning steps, which can lead to more accurate and logical conclusions** (Kong et al., 2023). Other work shows role-play can boost task reasoning (e.g., arithmetic/commonsense) by helping the model adopt a problem-solving stance (Zhao, et al., 2024). For example, embodying a "math teacher" persona encourages the LLM to adopt a more structured and explanatory approach to solving problems (Kong et al., 2023).
  
![Reasoning as a math teacher](Reasoning.png)

- **Caveats:** However, the introduction of personas has risks. Research has shown that while role-playing can improve reasoning, it can also **amplify inherent biases present in the training data**. Multiple studies report that role-play can amplify bias, toxicity, and unsafe outputs, especially when personas encode social attributes such as occupation, gender, race or religion (Shen et al., 2024).


**On Factual Accuracy:**

- While the results are mixed (sometimes task accuracy improves due to enhanced reasoning), larger studies and meta-analyses generally indicate that **adding a persona usually does not improve factual accuracy and can even slightly diminish it on average** (Zheng et al., 2024). The effects of personas are also highly unpredictable; the same role might assist on one item but hinder another, and simple strategies for selecting "the best role" perform almost randomly.
  
- **A Particular Caveat**: A 2024 study investigated how role-playing different social identities affects ChatGPT's ability to **classify misinformation**. The findings were stark: including social identities in prompts significantly reduced average detection accuracy from 68.1% (with no identity) to as low as 19.2% (with a political identity). This suggests that when a model adopts a persona with inherent biases or worldviews, its **capacity to objectively evaluate information for accuracy can be severely compromised** (Harris, 2024).



## Do the effects differ cross domains?
In specialized fields, context matters. For instance, in **medicine**, assigning an expert persona can improve the accuracy and comprehensiveness of patient-facing information (Li et al., 2024). 

Furthermore, a modeling approach called REGA (Wang et al., 2024) shows that when models are fine-tuned using role prompts in fields like law, medicine, or finance, and then used with a generalist role, their performance improves across these domains compared to baselines. This highlights a more integrated **training strategy — not a user-level prompt though**(Wang et al., 2024).

## A short conclusion
It appear that, in general, role-play prompting does not lead to better task outcomes/accuracy. Even if it does, it’s typically through improved **domain-structured reasoning** (e.g., clinical reasoning prompts), or by adopting a specific framework of thinking and reasoning in the particular field. But be careful — this could also lead to **enhanced bias** embedded in the training data. 


**When to Consider Using Role-Play Prompting**

- ***When You Need Structured, Domain-Specific Reasoning***: As mentioned, a relevant persona can better guide the model to adopt a particular domain-specific reasoning framework, accessing and applying specialized knowledge.

- ***When Tasks Are Subjective, Personal, or Creative***: Persona prompts truly excel in interpretive or creative tasks where a particular viewpoint, communication style, or emotional resonance is desired.

**Avoid Role-Play Prompting**

- ***When Seeking Factual Information***: Introducing a persona can reduce factual accuracy, making it less suitable for tasks requiring strict objectivity(we shouldn't rely on LLMs for exact factual knowledge anyways).

- ***When Concerned About Bias***: If the output's fairness, neutrality, or safety is paramount, assigning a persona, especially one with social attributes, risks amplifying problematic elements from the training data.
