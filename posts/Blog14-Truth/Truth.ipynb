We often talk about Artificial Intelligence as if it were a neutral arbiter—a cold, calculating machine that spits out objective facts. We feed it data, it learns the patterns, and it gives us the "truth." But reading Chapter 10, "Arbiter of Truth," in Unmasking AI stopped me in my tracks and forced me to ask a more fundamental question: **Is the very act of training an AI an act of bias?**

The author argues that training a machine learning model is, at its core, an exercise in setting standards. We are trying to teach a machine to recognize "ground truth." But as the book eloquently points out, truth is a slippery thing. It evolves. It shifts. And crucially, it never escapes the gravity of human prejudice.


## The Myth of the “Objective” Label

To understand this, we have to look at the mundane task of model training. Before a model can recognize a “spam email” or a “qualified job candidate,” either a human has to sit down and tell it what those things look like (supervised learning), or humans define what “success” looks like mathematically (deep learning). Usually, success means “minimizing error” across the majority of the data.

By doing these, they are codifying their own perspective, or freezing a specific moment of cultural consensus (or lack thereof) into a mathematical standard. If I label a resume as “high potential,” I am using my definition of potential, shaped by my background, my biases, and my blind spots. When we aggregate these labels, we aren’t creating a “universal truth”, but are creating a statistical average of our collective subjectivities.


## “Truth” is a Moving Target

The book makes a compelling philosophical point: **“Truth” is not static.** What we considered a medical “truth” fifty years ago might be considered malpractice today. What was considered a social “norm” a decade ago might be unacceptable now.
When we train an AI on historical data, we are essentially trapping it in the past. We are asking it to predict the future using the “truths” of yesterday—truths that might have been discriminatory, incomplete, or flat-out wrong. If we treat these models as objective authorities, we risk calcifying those old prejudices, giving them a veneer of scientific legitimacy they no longer deserve.

## Re-thinking the Goal

So, where does this leave us? If the decision-making process can never be truly "objective" because the inputs are soaked in human perspective, should we give up?
I don't think so. **But the public needs to stop looking at AI as a "Truth Machine." Instead, we should view it as a mirror.** It reflects our current standards, our current definitions, and yes, our current biases. The implications of this are practical and urgent:

1. ***Transparency in the “Standard-Setting”:*** We need to stop hiding the labeling process. We need "Datasheets for Datasets" that disclose who labeled the data, what instructions they were given, and what demographic perspectives were included. We need to know whose version of "truth" the model was taught.
2.  ***Model Expiration Dates:*** Since social truth evolves, AI models should not be permanent. A model trained on 2020 social data might be dangerously outdated by 2025. We need protocols that force models to be "sun-setted" or retrained to reflect shifting social norms and new knowledge.
3.  ***The Right to Contest:*** If an AI is enforcing a subjective standard—like "creditworthiness" or "employability"—subjectivity implies the possibility of error. Therefore, individuals must have a robust "right to contest" an AI's decision. We cannot hide behind the "black box" excuse; if the machine's "truth" hurts a human, the human must have the power to challenge that truth.


The danger isn't that models are biased but assuming they aren't. Once we accept that machine learning is fundamentally a process of subjective standard-setting, we can stop asking "Is this AI right?" and start asking the deeper question: "Whose truth is this AI enforcing, and is that a truth we still want to live by?"
