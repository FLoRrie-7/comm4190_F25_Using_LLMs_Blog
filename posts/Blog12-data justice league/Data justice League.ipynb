## A Book I'm Reading
I've been immersed in my "Living with a book" assignment recently, and the book I'm reading, Unmasking AI by Dr. Joy Buolamwini, is really interesting. The author at some point discusses the term "Justice League" and its deep roots in social equity campaigns, then recounts how a poster for the "Algorithmic Justice League" at an academic event deeply resonated with her. This reminds me of a curriculum project I designed at Penn GSE: The Data Justice League. The core idea was similar: **how do we make the abstract dangers of AI bias concrete, urgent, and understandable for young people?**

## Youth as Vulnerable AI Users
As an educator, I'm increasingly concerned about the younger generation—the "**AI natives**." They are growing up intensively using these tools, but often **lack the critical perspective** to perceive the potential harm beneath the surface. Dr. Buolamwini's book describes many vivid, real-world examples of how algorithmic bias causes serious harm, and it highlights a critical point: people who aren't personally affected by a specific bias often have no awareness that the issue even exists.


## Real-world Problems
This isn't a hypothetical problem. We're already seeing its consequences. A Stanford study, for example, found that automated systems often falsely flag essays written by non-native English speakers as AI-generated, creating a **high-stakes risk of false cheating accusations**. We also see Harmful Algorithmic Pathways on Social Media: Lawsuits filed against platforms like Instagram and TikTok have provided evidence of how algorithms, unaware of a user's vulnerability, can create harmful feedback loops. In one case, a young student who initially searched for fitness content was gradually shown more and more posts about eating disorders. The **algorithm's "engagement" bias contributed to the student developing anorexia and needing psychiatric treatment**.

## My project
Similar news shows why it is so pressing to unravel the underlying logic of AI. We cannot let children interact with these systems as if they are magical, all-knowing black boxes. This conviction drove my "Data Justice League" curriculum project. My goal was to make bias tangible for high schoolers using a hands-on activity with Teachable Machine. For instance, I had students train a model to identify "stray cats" using only data of clean, healthy house cats. Naturally, the model failed to recognize the injured, dirty stray cats it was tested on. This failure aims to provide a concrete "aha!" moment, sparking discussion on how biased data can lead AI to exclude or harm marginalized groups—the very ones most in need of help.

![Learning Goals of My Design](Curriculum.png)

Moving from abstract warnings to hands-on learning is the key. By providing tangible projects and real-world examples, we can empower children to interact with AI not just as passive consumers, but as critical, aware, and engaged citizens. Right now, fostering this AI literacy isn't just an add-on to our curriculum. It should be an **essential educational goal** in the era of AI.
