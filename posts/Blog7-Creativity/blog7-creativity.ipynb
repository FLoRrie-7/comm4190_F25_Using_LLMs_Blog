## The affective component of creativity - AI lacks this faculty?
The paper Next Token Prediction Is a Dead End for Creativity: Why It’s Impossible to Lose Yourself in the Moment was interesting to read. While many studies in creativity focus on the mere cognitive aspects of creativity such as convergent and divergent thinking, it emphasizes the importance of an essential affective component of creative generation, describing it as a “flow to lose oneself”. 


![The state of "flow"](flow.png)

I found this perspective inspiring, as it suggests that instantaneous creativity requires **tangible experience**, a unique **subjective viewpoint**, and even a **particular physiological state**. Language, at its core, is just an expression of our feelings and thoughts. Since LLMs lack physical faculties and a sense of "self," they can **only mimic a representation of creative products** in the form of language without drawing from these true, internal sources.


![If AI can subjectively experiencing emotions](emotions.png)

## Big data and the concept of "Average"
This idea reminded me of the concept of “Average” in the book [The End of Average: How We Succeed in a World That Values Sameness](https://www.toddrose.com/endofaverage) by Todd Rose. Influenced by the Industrial Revolution, various fields once sought to define an "average person"—educators designing standardized tests for comprehensive intelligence, medical institutions determining average health measurements, and even the Air Force attempting to design a cockpit for an average-sized pilot. However, they eventually realized that no real individual ever fit these average indices; **the concept of an "average person" proved largely meaningless**. 


![“An Average American"](average.png)

## An analogy: Are LLMs just "averaging" human thoughts?
The LLM's logic of predicting the probabilities of each token is conceptually analogous to this idea of "averaging." **Instead of carrying an individual's unique perspective, experience, and emotional state, LLMs seem to "average" human linguistic productions by predicting the next statistically plausible tokens**. Given the inherent randomness within these subjective human factors, how can such an "averaged" output ever be truly creative?


## Using LLMs for creative process facilitation
Yet, arguably, we can still utilize LLMs for creative production, or at least for facilitation. The instantaneous creativity discussed in the paper represents **only one form of improvised creativity**. Other important creative processes, such as **continuous iterations and development** on creative ideas, can certainly lead to highly novel outputs. These can often be achieved through strategic prompting. As the paper suggests, we should view LLMs as a **co-creative infrastructure** designed to support and enhance human ingenuity.
