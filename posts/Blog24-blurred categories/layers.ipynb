In casual summaries of the recent OpenAI report, a neat, quantifiable line often emerges: "Most people ask for advice, some do tasks, and ~11% is emotional chatting."

This narrative is compelling because it suggests a satisfying role for AI: the "therapist-lite."

But here is the subtle and critical research-methods flaw: The 11% figure represents the Expressing intent category, which is a functional classification, not the topic of "Relationships/Personal Reflection," which is much smaller (around 1.9%).

This confusion—mixing labels that live at **fundamentally different levels of analysis—compromises both interpretation and application.**


## Why This Matters for Education Research

In education and cognitive science, rigorous analysis demands clear separation of layers:

- Content Layer (The What): Topic (e.g., historical dates, friendship conflict).

- Action Layer (The Why/How): Intent (e.g., seeking advice, executing a task, venting).

- Evaluation Layer (The Outcome): Quality (e.g., user satisfaction, task success).

Mixing these layers generates methodologically false insights.

For instance: A student might use the Topic "Relationships/Personal Reflection." If their Intent is classified as Doing (requesting a formal apology letter draft), the system is being used for writing practice, not emotional catharsis. If you collapse that interaction into "emotional chat," you fundamentally misdiagnose the student's learning goal and the tool's effectiveness.

## A Practical Framework: Three-Layer Labeling for Integrity

To ensure methodological integrity in learning product design, every interaction must be logged across three critical, non-redundant layers:

- Intent (Functional Goal): What is the user attempting to achieve? Captures the overarching user aim, independent of topic. Examples: Seeking Advice (Asking), Performing a Task (Doing), Sharing Thoughts (Expressing).

- Topic (Domain Context): What domain is the conversation about? (Writing, Math, Relationships, Career).

- Learning Move (Cognitive Action): What specific cognitive operation is being requested? (Generate, Critique, Compare, Explain, Test, Reflect).

The Learning Move layer—focused on the cognitive process—is the most educationally significant and often the least captured by current product metrics.

Debunking myths like the "11% emotional chat" underscores **the need for layered analysis in AI research**. As the OpenAI report illustrates, precision in data interpretation is key to harnessing AI's transformative power for education.
