“Alignment” used to be a term I mostly heard in AI safety discussions: Does the model behave the way humans actually want? But in classrooms, the biggest risk isn’t always bias or hallucination. Sometimes the model is factually correct and still pedagogically wrong. It responds in a way that doesn’t match the learning goal, the teaching strategy, or the social setting. The technical term for that is contextual misalignment: the model fails to fully grasp the educational context and defaults to a teaching style that may be inappropriate for the moment. 

This idea is the core of a 2025 paper in Education and Information Technologies: *“Explainable pedagogical alignment: auditing LLMs with learning theories” (Domínguez Figaredo et al., 2025).*


The paper’s bold move: treat “teaching style” as an alignment target
Instead of asking whether GPT-4 is polite, safe, or accurate, the authors ask: Which learning theory does GPT-4 actually align with when it teaches? 

They use three classic learning frameworks as an “audit lens”:

- Behaviorism: structured instruction, explicit steps, feedback and reinforcement
- Active learning: prompting students to think, explain, explore, and construct understanding
- Social learning: learning through collaboration, peer interaction, and community activity 
This sounds theoretical—until you see what happens in a real tutoring platform.


## Study Results

***Study 1***: Students asked for active learning. GPT-4 answered like a behaviorist tutor.
The first study audits GPT-4 “in the wild” through Khanmigo on Khan Academy courses, tagging what students asked and what the model responded with. 

Here’s the headline result that stuck with me:
- Student prompts leaned Active Learning (the biggest slice: 50.94%)
  
- GPT-4 outputs leaned Behaviorism (the biggest slice: 59.32%) 

  In other words, even when students are trying to engage in inquiry (“help me reason through this”), the model often drifts into the safest, most structured mode (“here are the steps, here’s the answer, now practice”).
  
And when the course itself was designed around each theory, the gap became even clearer:

- Behaviorism alignment was strong: 71.99% of outputs matched behaviorist principles
  
- Active learning alignment was weaker: only 24.63%

- Social learning was weakest: only 10.74% ResearchGate

If you’ve ever felt that AI tutors are great at “explaining” but less great at “facilitating,” this is the quantitative version of that intuition.


## Why this matters (especially for EdTech)

A tutoring model that defaults to behaviorism isn’t bad. In fact, it’s extremely useful for procedural skill practice, immediate feedback, structured walkthroughs, exam prep and fluency building.

But it becomes a problem when the learning goal shifts to: conceptual understanding, metacognition (“why did you think that?”), student agency or group learning / discussion / peer critique.

A human teacher can adapt to the needs naturally: sometimes you lecture, sometimes you ask questions, sometimes you step back and let peers struggle together. The paper suggests GPT-4 doesn’t switch as gracefully. 

## Final Thoughts

We’ve spent a lot of time debating whether AI outputs are true or biased. This paper pushes a more uncomfortable question: even when the model is fluent and accurate, is it teaching in a way that actually serves the moment? “Contextual misalignment” is basically the education version of a subtle UI bug—the system is technically functioning, but it’s optimizing for the wrong objective.

The deeper implication is that pedagogical alignment isn’t a single setting. It’s a **dynamic matching problem**: matching learner state + task type + social context to an appropriate teaching move. Human teachers do this almost invisibly. They read signals (confusion vs curiosity), and they shift gears (active inquiry when the student has a foothold, social learning when meaning needs to be negotiated).

So the frontier isn’t “make the model act less behaviorist” in general. It’s building systems that can purposefully detect what kind of help is needed, and then route to the right teaching method.
