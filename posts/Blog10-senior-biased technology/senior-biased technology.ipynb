While the economic impacts of the pandemic are still lingering, the surge of AI is making the job market even more challenging—especially for new graduates and those in junior-level positions.

A new study, "Generative AI as Seniority-Biased Technological Change," analyzed a massive dataset of nearly 62 million U.S. résumés and 245 million job postings, providing one of the clearest pictures yet of what’s happening inside American companies. The researchers categorized firms as "AI-adopters" or "non-adopters" and compared hiring trends for junior versus senior positions.

## What the Study Found…

**1. The Post-2023 Divergence is Real**
   Before early 2023, the number of junior and senior employees at companies grew at similar rates. However, after the first quarter of 2023, the data shows a clear divergence: junior employment began to decline, while senior roles continued their upward trend. Specifically, firms embracing AI saw their junior workforce shrink by about 7.7% within a year and a half compared to similar companies not yet adopting AI. When comparing junior to senior roles within these AI-adopting firms, the drop for juniors was even more pronounced, at around 12%. This truly highlights the "Generative AI as Seniority-Biased" impact we're seeing.

![Recruitment trend disparity](figure3.png)

**2. It’s About Hiring Less, Not Firing More**
   The decline in junior positions isn't being driven by layoffs or resignations, but almost entirely by a significant slowdown in hiring new, entry-level talent. Firms adopting AI are hiring about 3.7 fewer junior employees per quarter—roughly a 22% reduction from previous levels. In fact, junior employee departures actually decreased slightly.


**3. The "U-Shaped" Impact on University Graduates**
    The impact also varies depending on a candidate's educational background. Graduates from **mid-tier universities (Tiers 2 and 3) are experiencing the most substantial reductions** in hiring. In contrast, those from **elite institutions (Tier 1) and lower-tier schools (Tiers 4 and 5) are less affected**, with the impact on graduates from the lowest-prestige institutions being statistically minimal. The paper interprets this as a **strategic cost-quality trade-off** for companies. Highly skilled graduates from top-tier schools might be too productive to replace. At the other end of the spectrum, the lowest-paid workers might be too inexpensive for AI substitution to be worthwhile. This dynamic leaves the broad "middle" of the junior talent pool facing the tightest squeeze.


## Final thoughts - the societal impact
On one hand, AI's transformation of the labor force presents economic opportunities, allowing companies to automate routine, entry-level tasks, which is a hallmark of technological progress. However, it's worth remembering that **all of today's seniors started as juniors.** They built their sophisticated judgment through the hands-on, real-world training that entry-level roles have always provided, and grew to become senior leaders who can effectively manage and leverage AI in their domain of expertise.

This surge in AI could become a significant barrier to young people’s growth and future development. What happens if an "experience gap" develops, where few young workers get the chance to be trained within their industry and grow into professionals capable of critically adopting and being empowered by AI? This societal challenge demands a **fundamental rethinking of how we design the education system, cultivate expertise, and transmit knowledge in an era where the traditional path from novice to expert is being somehow disrupted**.
