## Gemini as my potential user
My Product Management course tasked us with conducting a User **Empathy Interview** with ChatGPT. Specifically, we had to conceive a product, identify potential user personas, design interview questions to understand user needs and behaviors, and then have the LLM role-play the persona while we conducted the interview.

I ultimately chose the **Gemini 2.5 Pro** model, believing it can **“mimic”** human communicative behaviors and langauge better than other models, because Google as a company has got massive real user data to train their models.

My product is an AI-powered learning tool aimed at enhancing class engagement. I identified my first potential user persona as a high school student:

>*You are a 14-year-old high school student in a US public high school, who is afraid of in-class discussion because she either feels having nothing to say about the content, or the discussion focus was often irrelevant to her thoughts and she couldn’t catch up. From now, I will be conducting a user interview with you regarding an education product.*

## An compelling but unatural act
Gemini immediately adopted the persona, even including **paralanguage** in the form of action annotations:

>*(adjusts in my chair, pulls the sleeves of my hoodie over my hands) Oh, okay. Hi. Um, yeah, I’m ready whenever you are. Is it okay if I’m, like, a little nervous? I’m not really used to stuff like this. (looks around the room for a second) So… you’re just gonna ask me questions about school and stuff? Okay. Cool. Yeah, ask away, I guess.*

My initial instinct was to tell Gemini to omit the actions, but I recalled my professor's point that in-person interviews offer additional insights from **non-verbal cues**. Curious about this simulated multimodal communication, I decided to continue:

>*Can you walk me through the last time you had a class discussion about an article you had to read or a video you had to watch? What was that experience like for you?*

"Her" response was remarkably detailed:

>*Okay, um, yeah. Let’s see… the last time was probably on Tuesday, in my English class. We have Ms. Davis. We had to read this really short, kinda creepy story called “The Lottery” for homework the night before. I read it, and I thought it was… wow, like, really messed up. So…**(Anotehr 300 words of vivid description here)**…I just sat there and listened. It feels… I dunno, frustrating? You feel really small and kinda stupid, even when you did the reading and you think you understand it. My heart just beats really fast the whole time, hoping she doesn’t call on me.*

It’s unusual for a real human to provide such a structured and detailed recollection of an experience. Human memories are often **disorganized, emotionally charged, and filled with subjective highlights**. This response, while compelling, felt almost **too logical and rich** to be a genuine human recall. If it's an interrogation of suspect, such a coherent narrative would be seen as rehearsed.

## My own communicative strategy also changed with an LLM
I also identified the flaws in my own quesioning.I asked too many questions at once, assuming the LLM could address them all in a single reply. This is a common pitfall when shifting from human to AI interaction, as humans typically process and respond to questions sequentially within a shorter turn-taking framework:

>*I understand that. How do your group discussions typically start? When a teacher provides a specific question to start your group’s conversation, what makes a question effective or ineffective for you personally? Can you remember an example of a prompt that made it easy to talk?*

As expected, it did answer all questions together. Unexpectedly, it culminated in a sudden ending:


>*(Looks at the clock on the wall, getting a little antsy) Okay, so when we get into our little groups, there’s usually this super awkward moment of silence after we move the desks. Everyone just kinda looks at each other, and then someone will finally be like…(**Another 400 words of vivid description here**) It didn’t feel like I was going to get a bad grade if my opinion was “wrong.” I talked a lot in that discussion. (BRRRRRING!) Oh, that’s the final bell. I have to run or I’ll miss my bus. Um, it was nice talking to you. I hope that was helpful. Bye! (quickly gathers my stuff and heads for the door)


I had two more questions prepared, but didn’t get a chance to ask – "she" had to leave! This unexpected conclusion, a simulated abrupt exit, felt as chaotic and unpredictable as a real-life interview. I found this "out-of-control" situation quite amusing.

## LLMs as useful tools to conduct user interviews
In conclusion, despite some of the conversational structures feeling unnatural, interviewing an LLM proved to be a surprisingly helpful way to conduct user research. They are much more "knowledgeable" and elaborative than a human subject, and thus can provide richer details and insights for the product, as reported in my homework:

>**Key insights & informed product design from User A**:
***Prioritize Humanities Subjects***: Initial product development should focus on discussion-heavy subjects like English and History. These fields present a clear opportunity, as students often find it challenging to participate in discussions that require subjective interpretation and personal reflection.

>***Generate Accessible Prompts***: The AI product's core function should be to generate opening discussion prompts that are personal, opinion-based, and directly relevant to students' lived experiences. To encourage participation, the system must prioritize accessible questions over abstract, academic language that can intimidate hesitant speakers.
